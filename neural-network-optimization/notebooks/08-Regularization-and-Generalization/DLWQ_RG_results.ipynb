==========================================
Methods
==========================================

```
methods = [
    "baseline",
    "l1",
    "l2",
    "dropout",
    "earlystopping"
]
```

==========================================
Load Results
==========================================

```
all_results = []
```

```
for method in methods:

    df = pd.read_csv(
        results_dir /
        f"{method}_results.csv"
    )

    all_results.append(df)
```

```
comparison = pd.concat(
    all_results,
    ignore_index=True
)
```

```
comparison = comparison.sort_values(
    "RMSE"
)
```

```
comparison.to_csv(
    comparison_dir /
    "regularization_comparison.csv",
    index=False
)
```

```
print(
    "\n----- REGULARIZATION COMPARISON -----"
)
```

```
print(
    comparison
    .round(4)
    .to_string(index=False)
)
```

==========================================
Summary Export
==========================================

```
ranking = comparison.sort_values(
    by="RMSE"
)
```

```
with open(
    comparison_dir /
    "regularization_summary.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "Regularization and Generalization Study\n"
    )

    f.write(
        "=" * 50
    )

    f.write(
        "\n\nResults\n\n"
    )

    f.write(
        comparison
        .round(4)
        .to_string(index=False)
    )

    f.write(
        f"\n\nWinning Method: {best_method}"
    )

    f.write(
        "\n\nWinning Performance\n"
    )

    f.write(
        f"\nMAE: {winner['MAE']:.4f}"
    )

    f.write(
        f"\nRMSE: {winner['RMSE']:.4f}"
    )

    f.write(
        f"\nAdjusted R²: "
        f"{winner['Adjusted_R2']:.4f}"
    )

    f.write(
        f"\nTraining Time: "
        f"{winner['Training_Time_Seconds']:.2f} seconds"
    )

    f.write(
        f"\nGeneralization Gap: "
        f"{winner['Generalization_Gap']:.4f}"
    )

    f.write(
        f"\nFinal Training Loss: "
        f"{winner['Final_Training_Loss']:.4f}"
    )

    f.write(
        f"\nFinal Validation Loss: "
        f"{winner['Final_Validation_Loss']:.4f}"
    )

    f.write(
        "\n\nSelected Configuration Moving Forward\n"
    )

    f.write(
        "\n- Scaling Method: StandardScaler"
    )

    f.write(
        "\n- Activation Function: Sigmoid"
    )

    f.write(
        "\n- Architecture: 64 -> 32 -> 16"
    )

    f.write(
        "\n- Optimizer: RMSProp"
    )

    f.write(
        "\n- Epochs: 125"
    )

    f.write(
        "\n- Regularization: None (Baseline)"
    )

    f.write(
        "\n\nFinal Ranking (RMSE)\n"
    )

    for i, row in enumerate(
        ranking.itertuples(),
        start=1
    ):
        f.write(
            f"\n{i}. {row.Method}"
        )

    f.write(
        "\n\nGenerated Files\n"
    )

    f.write(
        "- regularization_comparison.csv\n"
    )

    f.write(
        "- mae_comparison.png\n"
    )

    f.write(
        "- rmse_comparison.png\n"
    )

    f.write(
        "- adjusted_r2_comparison.png\n"
    )

    f.write(
        "- training_time_comparison.png\n"
    )

    f.write(
        "- generalization_gap_comparison.png\n"
    )

    f.write(
        "- regularization_metrics_comparison.png\n"
    )

    f.write(
        "- regularization_learning_curve_comparison.png\n"
    )

    f.write(
        "- winner_learning_curve.png\n"
    )
```
