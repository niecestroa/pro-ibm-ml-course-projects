```
import pandas as pd
import matplotlib.pyplot as plt
```

==========================================
Directories
==========================================

```
results_dir = Path(
    r"C:\Users\aniec\Desktop\ibm-ml-project\deep-learning-rl\07-Optimizer-Benchmarking\tables_figures"
)
```

```
comparison_dir = Path(
    r"C:\Users\aniec\Desktop\ibm-ml-project\deep-learning-rl\07-Optimizer-Benchmarking\comparison"
)
```

```
comparison_dir.mkdir(
    parents=True,
    exist_ok=True
)
```

==========================================
Optimizers
==========================================

```
optimizers = [
    "sgd",
    "rmsprop",
    "adam"
]
```

==========================================
Load Results
==========================================

```
all_results = []
```

```
for optimizer in optimizers:

    df = pd.read_csv(
        results_dir /
        f"{optimizer}_results.csv"
    )

    all_results.append(df)
```

```
comparison = pd.concat(
    all_results,
    ignore_index=True
)
```

```
comparison = comparison.sort_values(
    by="RMSE"
)
```

==========================================
Export Comparison
==========================================

```
comparison.to_csv(
    comparison_dir /
    "optimizer_comparison.csv",
    index=False
)
```

```
print(
    "\n----- OPTIMIZER COMPARISON -----"
)
```

```
print(
    comparison
    .round(4)
    .to_string(index=False)
)
```

==========================================
Winner
==========================================

```
winner = comparison.iloc[0]
```

```
best_optimizer = winner["Optimizer"]
```

==========================================

```
ranking = comparison.sort_values(
    by="RMSE"
)
```

==========================================
Summary Export
==========================================

```
with open(
    comparison_dir /
    "optimizer_summary.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "Optimizer Benchmarking Study\n"
    )

    f.write(
        "=" * 50
    )

    f.write(
        "\n\nResults\n\n"
    )

    f.write(
        comparison
        .round(4)
        .to_string(index=False)
    )

    f.write(
        f"\n\nWinning Optimizer: {best_optimizer}\n"
    )

    f.write(
        "\nWinning Performance\n"
    )

    f.write(
        f"\nMAE: {winner['MAE']:.4f}"
    )

    f.write(
        f"\nRMSE: {winner['RMSE']:.4f}"
    )

    f.write(
        f"\nAdjusted R²: "
        f"{winner['Adjusted_R2']:.4f}"
    )

    f.write(
        f"\nTraining Time: "
        f"{winner['Training_Time_Seconds']:.2f} seconds"
    )

    f.write(
        f"\nFinal Training Loss: "
        f"{winner['Final_Training_Loss']:.4f}"
    )

    f.write(
        f"\nFinal Validation Loss: "
        f"{winner['Final_Validation_Loss']:.4f}"
    )

    f.write(
        "\n\nSelected Configuration Moving Forward\n"
    )

    f.write(
        "\n- Scaling Method: StandardScaler"
    )

    f.write(
        "\n- Activation Function: Sigmoid"
    )

    f.write(
        "\n- Architecture: 64 -> 32 -> 16"
    )

    f.write(
        "\n- Epochs: 125"
    )

    f.write(
        f"\n- Optimizer: {best_optimizer}"
    )

    f.write(
        "\n\nFinal Ranking (RMSE)\n"
    )

    for i, row in enumerate(
        ranking.itertuples(),
        start=1
    ):
        f.write(
            f"\n{i}. {row.Optimizer}"
        )

    f.write(
        "\n\nGenerated Files\n"
    )

    f.write(
        "- optimizer_comparison.csv\n"
    )

    f.write(
        "- mae_comparison.png\n"
    )

    f.write(
        "- rmse_comparison.png\n"
    )

    f.write(
        "- adjusted_r2_comparison.png\n"
    )

    f.write(
        "- training_time_comparison.png\n"
    )

    f.write(
        "- optimizer_metrics_comparison.png\n"
    )

    f.write(
        "- optimizer_learning_curve_comparison.png\n"
    )

    f.write(
        "- winner_learning_curve.png\n"
    )
```
