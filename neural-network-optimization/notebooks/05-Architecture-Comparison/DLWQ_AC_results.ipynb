==========================================
Load Results
==========================================

```
model_a = pd.read_csv(
    results_dir / "model_a_results.csv"
)
```

```
model_b = pd.read_csv(
    results_dir / "model_b_results.csv"
)
```

```
model_c = pd.read_csv(
    results_dir / "model_c_results.csv"
)
```

```
model_d = pd.read_csv(
    results_dir / "model_d_results.csv"
)
```

==========================================
Load Histories
==========================================

```
model_a_history = pd.read_csv(
    results_dir / "model_a_history.csv"
)
```

```
model_b_history = pd.read_csv(
    results_dir / "model_b_history.csv"
)
```

```
model_c_history = pd.read_csv(
    results_dir / "model_c_history.csv"
)
```

```
model_d_history = pd.read_csv(
    results_dir / "model_d_history.csv"
)
```

==========================================
RMSE Comparison (Best Model First)
==========================================

==========================================
Combine Results
==========================================

```
comparison = pd.concat(
    [model_a, model_b, model_c, model_d],
    ignore_index=True
)
```

```
comparison["Architecture"] = (
    comparison["Architecture"]
    .str.replace("_", " ")
)
```

```
comparison = comparison.sort_values(
    by="RMSE"
)
```

```
print("\n----- ARCHITECTURE COMPARISON -----")
print(comparison.round(4))
```

==========================================
Export Results
==========================================

```
comparison.to_csv(
    comparison_dir / "architecture_comparison.csv",
    index=False
)
```

==========================================
Determine Winner
==========================================

```
winner = comparison.iloc[0]["Architecture"]
```

```
print("\n----- BEST ARCHITECTURE -----")
print(winner)
```

==========================================
Export Summary
==========================================

```
with open(
    comparison_dir /
    "architecture_comparison_summary.txt",
    "w",
    encoding="utf-8"
) as f:

    f.write(
        "Architecture Comparison\n"
    )

    f.write(
        "=" * 50
    )

    f.write(
        "\n\nResults\n\n"
    )

    f.write(
        comparison.round(4).to_string(index=False)
    )

    winner_row = comparison.iloc[0]

    f.write(
        f"\n\nBest Architecture:\n{winner_row['Architecture']}"
    )

    f.write(
        f"\n\nWinning Performance"
    )

    f.write(
        f"\nMAE: {winner_row['MAE']:.4f}"
    )

    f.write(
        f"\nRMSE: {winner_row['RMSE']:.4f}"
    )

    f.write(
        f"\nAdjusted R²: {winner_row['Adjusted_R2']:.4f}"
    )
    
    f.write(
        "\n\nFinal Ranking\n"
    )

    for i, arch in enumerate(
        comparison["Architecture"],
        start=1
    ):
        f.write(
            f"\n{i}. {arch}"
        )
```
